## Histograms

To define the problem, we suppose that we have data $D = \left\{X_{j}\right\}_{j=1}^{N_{D}}$ where are i.i.d. and $X_{j}\sim p(x)$, where $p(x)$ is some unknown distribution.  A _histogram_, say $\hat{p}_{H}(x)$ is a model/estimator of $p(x)$.  

To construct it, if we know $x\in [a,b]$, we divide $[a,b]$ into $K$ bins of width $\delta x  = (b-a)/K$ so that 

$$
[a,b]=\cup_{l=1}^{K}[a_{l},b_{l}), ~ a_{1}=a, ~ b_{K} = b, ~b_{l}=a_{l+1}, ~ b_{l}-a_{l} = \delta x
$$ 

Define an _indicator_ function for a set $A$ as $I_{A}(x)$ so that 

$$
I_{A}(x) = \left\{
\begin{array}{rl}
1 & x \in A \\ 
0 & x \notin A
\end{array}
\right.
$$

Then $\hat{p}_{H}(x)$ is defined to be

$$
\hat{p}_{H}(x) = \frac{1}{\delta x} \sum_{l=1}^{K} \hat{p}_{l} I_{[a_{l},b_{l})}(x), ~ \hat{p}_{l} = \frac{N_{l}}{N_{D}}
$$

where $N_{l}$ is the number of data points in the $l^{th}$-bin, i.e.

$$
N_{l} = \sum_{j=1}^{N_{D}}I_{[a_{l},b_{l})}(X_{j}),
$$

so that $\sum_{l=1}^{K}N_{l} = N_{D}$.

Describing these histograms in terms of bias and variance, the first histogram (K = 2) clearly has very high bias since the super large bins just average out the values without showing important details about the data. It has lower variance since we have larger bins with a more stable structure and very little variability. The third one (K=200) is the opposite: it has low bias since it is closer to capturing the true distribution of the data, but high variance since it captures every single fluctuation and random noise due to the very high number of bins (and small bin size). The middle histogram sits right in the middle with low variance and bias: bin sizes are narrow enough to capture the distribution, but not too narrow to the point where we just capture random noise like in plot 3.

As we showed in class, the cross-validated error in using a $K$-bin histogram estimator is given by:

$$
E(K) = \frac{2}{\delta x (N_{D}-1)} - \delta x\frac{N_{D}+1}{N_{D}-1}\sum_{l=1}^{K}\left(\frac{\hat{p}_{l}}{\delta x}\right)^{2}
$$

Implement this and then explore how the cross-validated error changes as a function of bin count. Can you find an optimal bin count for the given data set?  

```python
def cross_val(bincnt, samples):
    nd = samples.size
    dx = (max(samples) - min(samples)) / bincnt # (b-a) / K
    hist, bins = np.histogram(samples, bins=bincnt, density=True)
    # note here that hist represents p-hat_l / dx, the estimated density in each bin
    error = (
        2 / (dx * (nd - 1))
        - dx * (nd + 1) / (nd - 1) * np.sum(hist**2) 
    )
    return bins, hist, error
```

```python
bincnts = np.arange(1, 100) # Explore bin counts 
errors = np.zeros(len(bincnts))
for jj, bincnt in enumerate(bincnts):
    bins, hist, error = cross_val(bincnt, vals2)
    errors[jj] = error
```

* Show that $\int_{a}^{b}\hat{p}_{H}(x)dx = 1$

We know already that:

$$
\hat{p}_{H}(x) = \frac{1}{\delta x} \sum_{l=1}^{K} \hat{p}_{l} I_{[a_{l},b_{l})}(x), ~ \hat{p}_{l} = \frac{N_{l}}{N_{D}}
$$

So plugging this in:

$$
\int_{a}^{b}\hat{p}_{H}(x) dx = \int_{a}^{b}\frac{1}{\delta x} \sum_{l=1}^{K} \hat{p}_{l} I_{[a_{l},b_{l})}(x) dx
= \frac{1}{\delta x} \sum_{l=1}^{K} \hat{p}_{l} \int_{a}^{b}I_{[a_{l},b_{l})}(x) dx
= \frac{1}{\delta x} \sum_{l=1}^{K} \frac{N_{l}}{N_{d}} \int_{a_l}^{b_l} 1 dx
= \frac{1}{\delta x} \sum_{l=1}^{K} \frac{N_{l}}{N_{d}} (b_l - a_l)
= \frac{1}{\delta x} \frac{N_{d}}{N_{d}} {\delta x}
= 1
$$
