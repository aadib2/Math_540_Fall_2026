```python
import numpy as np
from scipy.fft import fft
import matplotlib.pyplot as plt
import pickle
%matplotlib inline
```

```python
with open('data_set_1.pkl', 'rb') as f:
    data_set_1 = pickle.load(f)

with open('data_set_2.pkl', 'rb') as f:
    data_set_2 = pickle.load(f)

xvals1 = data_set_1[:, 0]
yvals1 = data_set_1[:, 1]

xvals2 = data_set_2[:, 0]
yvals2 = data_set_2[:, 1]
```

```python
def xp_mat_maker(xvals, p):
    xp = np.ones((xvals.size, p+1))
    for jj in range(p):
        xp[:, jj+1] = xvals * xp[:, jj]
    return xp
```

```python
def cos_mat_maker(xvals, freqs):
    cp = np.ones((xvals.size, len(freqs)))
    for jj, freq in enumerate(freqs):
        cp[:, jj] = np.cos(freq * xvals)
    return cp
```

```python
def least_squares_solve(mat, yvals):
    u, s, vt = np.linalg.svd(mat, full_matrices=False)
    alpha = (vt.T @ np.diag(1./s) @ u.T) @ yvals.reshape(-1, 1)
    error = np.linalg.norm(mat @ alpha - yvals.reshape(-1, 1))
    return alpha, error 
```

### 1.3

Since ${\bf C}^{-1}$ is the inverse of ${\bf C}$, we know that

$$
{\bf C}^{-1}{\bf C}={\bf I}_{n+m}.
$$

Using the partitioned forms of ${\bf C}^{-1}$ and ${\bf C}$, the left-hand side becomes

\begin{align*}
{\bf C}^{-1}{\bf C}
&=
\begin{pmatrix}
{\bf A}_{nn} & {\bf A}_{nm} \\
{\bf A}_{nm}^{T} & {\bf A}_{mm}
\end{pmatrix}
\begin{pmatrix}
{\bf C}_{nn} & {\bf C}_{nm} \\
{\bf C}_{nm}^{T} & {\bf C}_{mm}
\end{pmatrix} \\
&=
\begin{pmatrix}
{\bf A}_{nn}{\bf C}_{nn}+{\bf A}_{nm}{\bf C}_{nm}^{T}
& {\bf A}_{nn}{\bf C}_{nm}+{\bf A}_{nm}{\bf C}_{mm} \\
{\bf A}_{nm}^{T}{\bf C}_{nn}+{\bf A}_{mm}{\bf C}_{nm}^{T}
& {\bf A}_{nm}^{T}{\bf C}_{nm}+{\bf A}_{mm}{\bf C}_{mm}
\end{pmatrix}.
\end{align*}

The identity matrix has the corresponding block form

$$
{\bf I}_{n+m}=
\begin{pmatrix}
{\bf I}_{n} & {\bf O}_{nm} \\
{\bf O}_{mn} & {\bf I}_{m}
\end{pmatrix}.
$$

Equating corresponding blocks gives

\begin{align*}
{\bf A}_{nn}{\bf C}_{nn}+{\bf A}_{nm}{\bf C}_{nm}^{T} &= {\bf I}_{n}, \\
{\bf A}_{nn}{\bf C}_{nm}+{\bf A}_{nm}{\bf C}_{mm} &= {\bf O}_{nm}, \\
{\bf A}_{nm}^{T}{\bf C}_{nn}+{\bf A}_{mm}{\bf C}_{nm}^{T} &= {\bf O}_{mn}, \\
{\bf A}_{nm}^{T}{\bf C}_{nm}+{\bf A}_{mm}{\bf C}_{mm} &= {\bf I}_{m}.
\end{align*}

The third equation is the transpose of the second equation, so the three distinct identities requested are

\begin{align*}
{\bf A}_{nn}{\bf C}_{nn}+{\bf A}_{nm}{\bf C}_{nm}^{T} &= {\bf I}_{n}, \\
{\bf A}_{nn}{\bf C}_{nm}+{\bf A}_{nm}{\bf C}_{mm} &= {\bf O}_{nm}, \\
{\bf A}_{nm}^{T}{\bf C}_{nm}+{\bf A}_{mm}{\bf C}_{mm} &= {\bf I}_{m}.
\end{align*}


### 1.4

Starting with the second identity from Part 1.3,

$$
{\bf A}_{nn}{\bf C}_{nm}+{\bf A}_{nm}{\bf C}_{mm}={\bf O}_{nm},
$$

we move the first term to the right-hand side and multiply by ${\bf C}_{mm}^{-1}$ on the right. This gives

\begin{align*}
{\bf A}_{nm}{\bf C}_{mm}
&=-{\bf A}_{nn}{\bf C}_{nm}, \\
{\bf A}_{nm}
&=-{\bf A}_{nn}{\bf C}_{nm}{\bf C}_{mm}^{-1}.
\end{align*}

Next, using the third identity from Part 1.3,

$$
{\bf A}_{nm}^{T}{\bf C}_{nm}+{\bf A}_{mm}{\bf C}_{mm}={\bf I}_{m},
$$

we isolate ${\bf A}_{mm}$:

\begin{align*}
{\bf A}_{mm}{\bf C}_{mm}
&={\bf I}_{m}-{\bf A}_{nm}^{T}{\bf C}_{nm}, \\
{\bf A}_{mm}
&=\left({\bf I}_{m}-{\bf A}_{nm}^{T}{\bf C}_{nm}\right){\bf C}_{mm}^{-1}.
\end{align*}

Finally, start with the first identity from Part 1.3 and substitute the expression for ${\bf A}_{nm}$:

\begin{align*}
{\bf A}_{nn}{\bf C}_{nn}+{\bf A}_{nm}{\bf C}_{nm}^{T}
&={\bf I}_{n}, \\
{\bf A}_{nn}{\bf C}_{nn}-{\bf A}_{nn}{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}
&={\bf I}_{n}, \\
{\bf A}_{nn}\left({\bf C}_{nn}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}\right)
&={\bf I}_{n}.
\end{align*}

Therefore,

$$
{\bf A}_{nn}
=\left({\bf C}_{nn}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}\right)^{-1}.
$$

To put this in the requested form, factor ${\bf C}_{nn}$ on the right:

\begin{align*}
{\bf C}_{nn}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}
&=\left({\bf I}_{n}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}{\bf C}_{nn}^{-1}\right){\bf C}_{nn}.
\end{align*}

Using $({\bf A}{\bf B})^{-1}={\bf B}^{-1}{\bf A}^{-1}$, we obtain

\begin{align*}
{\bf A}_{nn}
&=\left[\left({\bf I}_{n}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}{\bf C}_{nn}^{-1}\right){\bf C}_{nn}\right]^{-1} \\
&={\bf C}_{nn}^{-1}\left({\bf I}_{n}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}{\bf C}_{nn}^{-1}\right)^{-1}.
\end{align*}

Thus, the desired formulas are

\begin{align*}
{\bf A}_{nn} &= {\bf C}_{nn}^{-1}\left({\bf I}_{n}-{\bf C}_{nm}{\bf C}_{mm}^{-1}{\bf C}_{nm}^{T}{\bf C}_{nn}^{-1}\right)^{-1}, \\
{\bf A}_{nm} &= -{\bf A}_{nn}{\bf C}_{nm}{\bf C}_{mm}^{-1}, \\
{\bf A}_{mm} &= \left({\bf I}_{m}-{\bf A}_{nm}^{T}{\bf C}_{nm}\right){\bf C}_{mm}^{-1}.
\end{align*}

**Problem 2**: Using a least-squares approach, find a polynomial model of the data set represented by `xvals1` and `yvals1`, i.e. find $p$ and corresponding $\alpha_{l}$ such that 

$$
y_{j} \approx \sum_{l=0}^{p}\alpha_{l}x_{j}^{l}
$$

Please explain how you arrived at your choice for $p$.  What is the condition number of your matrix $X_{p}$?  How does this value influence your choice of $p$?  What role does regularization play in improving your results?  


So I decided to set my condition #, p, to be 4 as it resulted in a better fit and smaller error. When playing around with surrounding values of 3,5, and 6 I got similar residuals. The condition number is ~2274. This value increases as we also increase the value of $p$, which results in the alphas or parameters having a much larger spread due to larger conditioning.
```python
pval = 4
xp = xp_mat_maker(xvals1, pval)
alpha, error = least_squares_solve(xp, yvals1)
print(f"Residual/error: {error}")
print(f"Condition Number: {np.linalg.cond(xp)}")
```

**Problem 3**: Using a least-squares approach, find a cos model of the data set reprsented by `xvals2` and `yvals2`.  In other finds, find $\alpha_{l}$ and $\omega_{l}$ such that

$$
y_{j} \approx \sum_{l=0}^{p}\alpha_{l}\cos(\omega_{l}x_{j}).
$$

To figure out the frequencies, you need to use a Fast Fourier Transform (FFT; see Book/Notes).  Then you find the $\alpha_{l}$ terms using least-squares fitting.  Explain your choice of frequencies and the quality of fit that this gives you to your data set.  How would you regularize your fit?  

I chose the frequencies $\omega = [0, 3, 8]$ because the FFT plot shows its largest low-frequency peaks at approximately $0$, $3$, and $8$. The zero-frequency term represents the mean or constant part of the signal. The other two frequencies represent the strongest oscillatory components in the displayed range.

For these frequencies, the matrix:

$$
C_{jl} = \cos(\omega_l x_j),
$$

is formed and $C\alpha \approx y$ is solved by least squares. The residual norm was approximately $158.6$.

To regularize the fit, you would apply the same expression as we did above. Instead of minimizing only the residual, you'd have to solve

$$
\min_{\alpha}\left(\|C\alpha-y\|_2^2 + \lambda\|\alpha\|_2^2\right),
$$

where $\lambda > 0$ controls the penalty on large coefficients. The solution is

$$
\hat{\alpha} = (C^T C + \lambda I)^{-1}C^T y.
$$

which is another way to write the solution from class, but doing so not in terms of the S.V.D. terms of matrix $C$.

This reduces the effect of noise and ill-conditioning
