**Problem 2**: We will now explore solving $A{\bf x}={\bf b}$ numerically.  First fill in the code skeleton below.

```python
Amat = np.zeros((3, 2))
bvec1 = np.zeros(3)
bvec2 = np.zeros(3)

Amat[0, 0] = 1
Amat[0, 1] = 3
Amat[1, 0] = 3
Amat[1, 1] = 2
Amat[2, 0] = 1
Amat[2, 1] = 2

bvec1[0] = 1
bvec1[1] = 10
bvec1[2] = 2

bvec2[0] = 1
bvec2[1] = 10
bvec2[2] = 1

sol1= np.linalg.lstsq( Amat, bvec1 , rcond=None)
sol2 = np.linalg.lstsq( Amat, bvec2 , rcond=None)
```

```python
print(f"The solution for b={[val for val in bvec1]} is x={sol1[0]}, {sol1[1]}")
print(f"The solution for b={[val for val in bvec2]} is x={sol2[0]}, {sol2[1]}")
```

The command `lstsq` found a solution to a problem that isn't supposed to have one.  What problem do you think it's actually solving? 

In this case `lstsq` solves this set of linear equations using the least squares method, finding the best possible solution by minimizing the error which in this case is the euclidean 2-norm of the residual vector: $\|b - Ax\|$. It found the solution to the second problem (which has more equations than unknowns) through finding the x values which minimized our objective. iF we were to examine the residuals from above, one would be non-zero and the other would be zero.


**Problem 3**: For $A:\mathbb{C}^{n}\rightarrow \mathbb{C}^{m}$, show that $\left<A{\bf x},{\bf y} \right> = \left<{\bf x}, A^{\dagger}{\bf y} \right>$, where $A^{\dagger}_{ij} = A^{\ast}_{ji}$ and 

$$
\left(A{\bf x}\right)_{l} = \sum_{j=1}^{n}A_{lj}x_{j}, ~ l=1, \cdots, m.
$$


**Solution**:

\begin{aligned}
\langle A\mathbf{x}, \mathbf{y} \rangle
&= \sum_{l=1}^{m} (A\mathbf{x})_l^* y_l \\[4pt]
&= \sum_{l=1}^{m}
\left(
\sum_{j=1}^{n} A_{lj}x_j
\right)^* y_l \\[4pt]
&= \sum_{l=1}^{m}
\sum_{j=1}^{n}
A_{lj}^* x_j^* y_l \\[4pt]
&= \sum_{j=1}^{n}
x_j^*
\left(
\sum_{l=1}^{m} A_{lj}^* y_l
\right) \\[4pt]
&= \sum_{j=1}^{n}
x_j^*
\left(
\sum_{l=1}^{m} A^\dagger_{jl} y_l
\right)
\qquad \text{since } A^\dagger_{jl} = A_{lj}^* \\[4pt]
&= \sum_{j=1}^{n}
x_j^*(A^\dagger \mathbf{y})_j \\[4pt]
&= \langle \mathbf{x}, A^\dagger \mathbf{y} \rangle.
\end{aligned}
