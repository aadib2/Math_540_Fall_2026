**Problem 1**: For a $n\times n$ unitary matrix $U$, so that $U^{-1}=U^{\dagger}$ show that 

$$
\left|\left|U{\bf x}\right|\right| = \left|\left|{\bf x}\right|\right|, ~ {\bf x}\in \mathbb{C}^{n}
$$

Please explain in words what this means.  


To do this, we can also show that $\left|\left|U{\bf x}\right|\right|^2 = \left|\left|{\bf x}\right|\right|^2$

$$



$$

To prove this, it is enough to show that the squared norms are equal:
$$
\|U\mathbf{x}\|_2^2
= \langle U\mathbf{x}, U\mathbf{x}\rangle
= (U\mathbf{x})^\dagger(U\mathbf{x}).
$$

Since the conjugate transpose reverses multiplication,
$$
(U\mathbf{x})^\dagger
= \mathbf{x}^\dagger U^\dagger.
$$

Therefore,
$$
\begin{aligned}
\|U\mathbf{x}\|_2^2
&= \mathbf{x}^\dagger U^\dagger U\mathbf{x} \\
&= \mathbf{x}^\dagger I\mathbf{x} \\
&= \mathbf{x}^\dagger\mathbf{x} \\
&= \|\mathbf{x}\|_2^2.
\end{aligned}
$$

Taking the nonnegative square root gives
$$
\boxed{\|U\mathbf{x}\|_2 = \|\mathbf{x}\|_2}.
$$

**Problem 2**: For a $n\times n$ self adjoint matrix $A$ (i.e. $A^{\dagger}=A$) show that all of its eigenvalues are real.  *Hint*, start from 

$$
A{\bf v} = \lambda {\bf v} \rightarrow \left<A{\bf v},{\bf v}\right> = \lambda \left|\left|{\bf v}\right|\right|^{2}_{2}
$$

Suppose that $\mathbf{v}\neq \mathbf{0}$ is an eigenvector of A
with eigenvalue $\lambda$, so that

$$
A\mathbf{v}=\lambda\mathbf{v}.
$$

Using the inner product from the hint,
$$
\langle A\mathbf{v},\mathbf{v}\rangle
= \lambda\|\mathbf{v}\|_2^2.
$$

Since $A$ is self-adjoint, $A^\dagger=A$. Therefore,
$$
\langle A\mathbf{v},\mathbf{v}\rangle
= \langle \mathbf{v},A\mathbf{v}\rangle.
$$

By conjugate symmetry of the inner product,
$$
\langle \mathbf{v},A\mathbf{v}\rangle
= \langle A\mathbf{v},\mathbf{v}\rangle^*
= \lambda^*\|\mathbf{v}\|_2^2.
$$

Combining these equalities gives
$$
\lambda\|\mathbf{v}\|_2^2
= \lambda^*\|\mathbf{v}\|_2^2.
$$

Because $\mathbf{v}\neq\mathbf{0}$, we have
$\|\mathbf{v}\|_2^2>0$. Thus,
$$
\lambda=\lambda^*,
$$
which means that $\lambda$ is real. Therefore, every eigenvalue of a
self-adjoint matrix is real.

**Problem 3**: For $n\times n$ self adjoint matrix $A$, show that if it is also positive semi-definite so that 

$$
\left<A{\bf x},{\bf x}\right> \geq 0 ~ \forall {\bf x}\in\mathbb{C}^{n}
$$

then all of its eigenvalues are non-negative.  To do this, just use the formula written in Problem 2.  


For this problem, let $\mathbf{v} \neq \mathbf{0}$ be an eigenvector of $A$ with eigenvalue
$\lambda$. Then
$$
A\mathbf{v} = \lambda\mathbf{v}.
$$

From Problem 2, we have
$$
\left\langle A\mathbf{v},\mathbf{v}\right\rangle
= \lambda\|\mathbf{v}\|_2^2.
$$

Because $A$ is positive semidefinite,
$$
\left\langle A\mathbf{v},\mathbf{v}\right\rangle \geq 0.
$$
Therefore,
$$
\lambda\|\mathbf{v}\|_2^2 \geq 0.
$$

Since $\mathbf{v}\neq\mathbf{0}$, its squared Euclidean norm is strictly
positive:
$$
\|\mathbf{v}\|_2^2
= \sum_{j=1}^{n}|v_j|^2 > 0.
$$
Thus, the only way for
$$
\lambda\|\mathbf{v}\|_2^2 \geq 0
$$
to hold is for
$$
\lambda \geq 0.
$$
Hence, all eigenvalues of $A$ are non-negative.

**Problem 5**: For given real-valued, $n\times n$ square matrix $A$ with SVD $A=U\Sigma V^{T}$, if we order the singular values $\sigma_{j}$ such that $\sigma_{1}\geq \sigma_{2}\geq \cdots \geq \sigma_{n}$, then we define its *condition number*, say $c(A)$, to be 

$$
c(A) = \frac{\sigma_{1}}{\sigma_{n}}
$$

Note, if $A$ is not of full rank then $\sigma_{n}=0$ and $c(A)=\infty$.

For $2 \leq n \leq 30$, using `np.random.randn` which generates random matrices using a normal distribution, find the average condition number as a function of $n$.  Plot your results and comment upon them.  Use the code skeleton below to get started.  

```python
nvals = np.arange(2,31) # range of n-vals
nsamples = 20 # how many different random samples per dimension do we want to generate?  how do your results change as you change this number?
average_condition_numbers = np.zeros(nvals.size)
for cnt, nval in enumerate(nvals):
    tot = 0
    for jj in range(nsamples):
        Amat = np.random.randn(nval, nval)
        # find the svd of Amat
        U, S, VT = np.linalg.svd(Amat)

        smallest_sigma = S[-1] # sigma at n

        if smallest_sigma > 1e-16: # don't include matrices that don't have full rank
            # update your average here
            current_condition_number = S[0] / smallest_sigma
            average_condition_numbers[cnt] += current_condition_number
            # keep track of how many terms you're including in your average
            tot += 1

    if tot > 0:   
        average_condition_numbers[cnt] /= tot # make sure you compute an average

plt.plot(nvals, average_condition_numbers)
plt.xlabel(r"$n$")
plt.ylabel("Average condition number")
plt.show()
```

Comment: With nsamples = 15 per value of n, we can see that the condition number fluctuates a lot as the size of the square matrix increases (maybe due to sampling noise). But on average it increases as n increases which means the largest singular value (the first one) tends to increase whilst the smallest one (the last) tends to become smaller as n increases.
