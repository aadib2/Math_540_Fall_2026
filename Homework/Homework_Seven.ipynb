**Problem 1**: In $\mathbb{R}^{n}$ with a fixed vector ${\bf a}$ describe the following sets.  Illustrate your descriptions via examples in $n=2$ and $n=3$.  Remember that for real vectors 
$$
\left<{\bf a},{\bf x}\right>=\left<{\bf x},{\bf a}\right>={\bf x}^{T}{\bf a}={\bf a}^{T}{\bf x}.
$$
  * $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>=0\right\}$
  * $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>\geq 0\right\}$
  * $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>= c, ~c > 0\right\}$
  * $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right> \geq c, ~c > 0\right\}$


1. $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>=0\right\}$: This set describes all vectors $x \in \mathbb{R}_n$ that are orthogonal to the fixed vector $\bf{a}$. Effectively it is the null space of $a^T$: $\mathbb{N}(a^T)$. Geometrically (seen in the scans attached in the drive) we can see when n=2, this just becomes a line orthogonal to a, and when n=3 it becomes a plane orthogonal to a.

2. $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>\geq 0\right\}$: This set describes all vectors $\mathbf{x}\in\mathbb{R}^n$ whose inner product with $\mathbf{a}$ is nonnegative, meaning they form an angle of at most $90^\circ$ with $\mathbf{a}$ (excluding the zero vector, for which the angle is undefined). Geometrically, this forms a closed half-space bounded by the hyperplane $\mathbf{a}^T\mathbf{x}=0$. For $n=2$, this is a half-plane, while for $n=3$, it is a half-space bounded by a plane, both including their boundaries. This is illustrated in the scan.

3. $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right>= c, ~c > 0\right\}$: This set describes all vectors whose inner product with $\mathbf{a}$ equals a fixed positive constant $c$. Geometrically, this forms a hyperplane perpendicular to $\mathbf{a}$, shifted a distance $c/|\mathbf{a}|$ from the origin in the direction of $\mathbf{a}$. For $n=2$, this is a line, while for $n=3$, it is a plane, neither of which passes through the origin.

4. $\left\{{\bf x}\in \mathbb{R}^{n} : \left<{\bf a},{\bf x}\right> \geq c, ~c > 0\right\}$: This set describes all vectors whose inner product with $\mathbf{a}$ is at least $c$. Geometrically, this forms a closed half-space bounded by the shifted hyperplane $\mathbf{a}^T\mathbf{x}=c$, including the boundary itself. For $n=2$, this is a half-plane, while for $n=3$, it is a half-space bounded by a plane, with the region extending in the direction of $\mathbf{a}$.

**Problem 3**: Momentum to Accelerate Convergence

So we can imagine that as the steepest descent method starts to converge onto a minimum of $f({\bf x})$, it faces a landscape in which $\nabla f({\bf x})$ doesn't change much from iteration to iteration.  To take advantage of this, we add what is called *momentum* to our steepest descent update formula.  This looks like the following algorithm.

\begin{align*}
\Delta {\bf x}_{k+1} =  & a\Delta {\bf x}_{k} - \mu \nabla f({\bf x}_{k})\\
{\bf x}_{k+1} = & {\bf x}_{k} + \Delta {\bf x}_{k+1}.
\end{align*}

Here, $0<a<1$ is the *momentum term* and we now fix the step $\mu$ to be a constant (though we still need to make sure we're taking descent directions so we need to be careful about choices for this parameter).  Since $\mu$ is now a constant, we call it the *learning rate* of our descent method.    

* Show that over some **very large** number of iterations such that $\nabla f({\bf x}_{k})\approx {\bf c}$, we can solve for $\Delta {\bf x}$ so that
$$
\Delta {\bf x} \approx -\frac{\mu}{1-a}{\bf c}
$$


Given $\nabla f({\bf x}_{k})\approx {\bf c}$, we can rewrite $\Delta {\bf x}$ as:

$$
\Delta {\bf x}_k =  a\Delta {\bf x}_{k-1} - \mu{\bf c}
$$

This then becomes:

$$
\begin{align*}
a(a\Delta{\bf x}_{j-2} - \mu{\bf c}) - \mu{\bf c}\\
= a^2 \Delta{\bf x}_{j-2} - (1+a)\mu{\bf c}\\
= a^2 \Delta{\bf x}_{j-2} - (a^0+a^1)\mu{\bf c}
\end{align*}
$$ 

Seeing that this follows the structure of a geometric series

$$
\begin{align*}
= a^j \Delta{\bf x}_0 - (1-a^j)\mu{\bf c}\\
= a^j \Delta{\bf x}_0 - \left(\frac{1-a^j}{1-a} \right)\mu{\bf c}
= \frac{-\mu}{1-a}{\bf c}
\end{align*}
$$




* What effect does the momentum then have on your method?  Why does adding momentum in effect automatically make your method have an adaptive step-size?  

So essentially, we can see that in the cases where the gradient doesn't change significantly and is roughly constant, momentum expands the step size via dividing $\mu$ by $1-a$ so that we try to converge a bit faster. Otherwise, we follow the normal gradient descent logic by stepping the size of $\mu$.


* To implement this method, we suppose that $\Delta {\bf x}_{0}=0$ and we then make a choice of $a$ and $\mu$.  Code this up and explore how the iteration counts can be improved by using different choices of $\mu$ and $a$.  

As a test function, we use

$$
f({\bf x}) = 2e^{x_{1}-.1}\cosh(3x_{2}) + e^{-x_{1}-.1}, ~ {\bf x} = (x_{1}, x_{2})
$$

One can quickly show that this has a global minimum at ${\bf x}_{\ast} = (-.5\ln(2), 0)$, and $f({\bf x}_{\ast})=2.5592666966582156$

```python
def f(x):
    x1, x2 = x
    return 2 * np.exp(x1 - 0.1) * np.cosh(3 * x2) + np.exp(-x1 - 0.1)

def grad_f(x):
    x1, x2 = x

    grad_x1 = (
        2 * np.exp(x1 - 0.1) * np.cosh(3 * x2)
        - np.exp(-x1 - 0.1)
    )
    grad_x2 = 6 * np.exp(x1 - 0.1) * np.sinh(3 * x2)

    return np.array([grad_x1, grad_x2])

def steepest_descent_with_momentum_f(x0, a, mu, tol, max_iter=100000):
    # code up logic here for this and the update steps
    x = np.array(x0, dtype=float)
    delta = np.zeros_like(x) # assume delta_0 = 0

    for iter in range(max_iter):
        gradient = grad_f(x)

        if np.linalg.norm(gradient) <= tol:
            return x, f(x), iter

        # update step
        delta = a * delta - mu * gradient
        x += delta

    raise RuntimeError("Gradient descent did not converge.")
```

```python
# implement the steepest descent with some toy data for x
x_star, f_star, iterations = steepest_descent_with_momentum_f(
    x0=[0.0, 1.0],
    a=0.5, # alpha
    mu=0.01, # lr
    tol=1e-14,
)

print(x_star)
print(f_star)
print(iterations)
```

```python
# let's plot how the number of iteration counts is impacted by different values of alpha
lrs = [0.01, 0.001]
alphas = [0.5, 0.6, 0.7, 0.8, 0.9]
x0 = [0.0, 1.0]

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

i=0
for lr in lrs:
    iters = []
    for a in alphas:
        _, _, iter = steepest_descent_with_momentum_f(x0, a=a, mu = lr, tol=1e-12)
        iters.append(iter)
    
    ax[i].plot(alphas, iters)
    ax[i].set_title(f"Learning Rate = {lr}")
    i+=1
        
plt.tight_layout()
plt.show()
```

We can see here that for the case when $\mu$ = 0.01, the # of iterations to converge decreases as alpha increases (makes sense since this increases the impact of the momentum technique), however when alpha becomes too large (0.8-0.9) it has the opposite effect and the gradient descent is likely to overshoot the minimum, increasing the number of iterations. When $\mu = 0.001$, since this is more conservative, we see a decreasing linear relationship between alpha and the number of iterations, which shows the value of the momentum term in helping us converge faster.

```python
# LOAD OUR Data

import kagglehub
data_path = kagglehub.dataset_download('harsh812saini/heart-disease-prediction-dataset')
df = pd.read_csv(data_path + '/Dataset.csv')

print(df.keys())
gender = df['gender']
male = gender == 'Male'
female = ~male

# modifying original code to store female data
features = [
    "age",
    "cholesterol (mg/dl)",
    "systolic B.P.(mm_hg)",
]

# define female DF after converting to numbers and removing nulls
female_df = df.loc[female, features].apply(
    pd.to_numeric,
    errors="coerce",
)

# Keep the rows that have valid values for every feature
valid_rows = female_df.notna().all(axis=1)
female_df = female_df.loc[valid_rows]

# Extract the feature arrays
age_raw = female_df["age"].to_numpy(dtype=float)
chol_raw = female_df["cholesterol (mg/dl)"].to_numpy(dtype=float)
systolic_bp_raw = female_df["systolic B.P.(mm_hg)"].to_numpy(dtype=float)

# Normalize
age = (age_raw - age_raw.mean()) / age_raw.std()
chol = (chol_raw - chol_raw.mean()) / chol_raw.std()
systolic_bp = (
    (systolic_bp_raw - systolic_bp_raw.mean())
    / systolic_bp_raw.std()
)

# Select outcomes using the exact same retained row indices
outcomes = df.loc[female_df.index, "Resting ECG Result"].to_numpy()
```

```python
#  erify there are no nulls (was an issue at first)
print(np.isnan(age).any())
print(np.isnan(chol).any())
print(np.isnan(systolic_bp).any())
```

```python
# define y
y = np.zeros(len(age))
healthy = outcomes == 'Normal sinus rhythm'
unhealthy = ~healthy
y[healthy] = 1
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection="3d")

ax.scatter(
    age_raw[healthy],
    chol_raw[healthy],
    systolic_bp_raw[healthy],
    color="blue",
    label="Healthy",
)

ax.scatter(
    age_raw[unhealthy],
    chol_raw[unhealthy],
    systolic_bp_raw[unhealthy],
    color="orange",
    label="Unhealthy",
)

ax.set_xlabel("Age")
ax.set_ylabel("Cholesterol (mg/dL)")
ax.set_zlabel("Systolic blood pressure (mm Hg)")
ax.set_title("Age vs. Cholesterol vs. Systolic Blood Pressure")
ax.legend()

plt.show()
```

```python
# create train and test datasets then train our model via GD
percentage_training = .9
num_men = age.size 
num_train_data = int(np.floor(percentage_training * num_men))
train_indices = np.sort(np.random.choice(np.arange(num_men), num_train_data, replace=False))
test_indices = np.setdiff1d(np.arange(num_men), train_indices)

Xtrain = np.zeros((3, num_train_data))
Xtrain[0, :] = age[train_indices]
Xtrain[1, :] = systolic_bp[train_indices]
Xtrain[2, :] = chol[train_indices]
ytrain = y[train_indices]

Xtest = np.zeros((3, num_men-num_train_data))
Xtest[0, :] = age[test_indices]
Xtest[1, :] = systolic_bp[test_indices]
Xtest[2, :] = chol[test_indices]
ytest = y[test_indices]

# fit to training data using logistic regression and GD for finding optimal params
w, beta = steepest_descent_with_momentum(ytrain, Xtrain, 0.99, 1e-3, 1e-14)
print(w, beta)
```

```python
# apply testing set with these params
test_hyperplane = Xtest.T @ w + beta * np.ones_like(ytest)
test_sig = 1 / (1 + np.exp(-test_hyperplane))
predictions = np.zeros_like(ytest)
healthy_prediction = test_sig >= 0.5
predictions[healthy_prediction] = 1
```

```python
# calculate the confusion matrix
true_negatives = np.sum((predictions == 0) & (ytest == 0))
false_positives = np.sum((predictions == 1) & (ytest == 0))
false_negatives = np.sum((predictions == 0) & (ytest == 1))
true_positives = np.sum((predictions == 1) & (ytest == 1))

c11 = true_negatives/np.sum(predictions == 0)
c12 = false_positives/np.sum(predictions == 1)
c21 = false_negatives/np.sum(predictions == 0)
c22 = true_positives/np.sum(predictions == 1)

class_confusion_matrix = np.array([[c11, c12], [c21, c22]])

print(class_confusion_matrix)
```

The performance here compared to the results with the male are that we are able to detect those that are unhealthy a bit worse (0.85 vs 0.76), but are able to detect those that are healthy a bit better (0.59 vs 0.64).
